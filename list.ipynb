```python
numbers = []

for count in range(20):
    value = int(input(f"Enter number {count + 1}: "))
    numbers.append(value)

print("\nNumbers entered:", numbers)
```

```python
print("\nRepeated values and their positions:")

for value in numbers:
    if numbers.count(value) > 1:
        positions = []
        for index in range(len(numbers)):
            if numbers[index] == value:
                positions.append(index)

        print(value, "-->", *positions)
```

```python
even = 0
odd = 0

for value in numbers:
    if value % 2 == 0:
        even += 1
    else:
        odd += 1

print("\nEven numbers:", even)
print("Odd numbers:", odd)
```

```python
numbers = []

for count in range(20):
    value = int(input(f"Enter number {count + 1}: "))
    numbers.append(value)

positive = 0
negative = 0

for value in numbers:
    if value > 0:
        positive += 1
    elif value < 0:
        negative += 1
    else:
        print("Zero found at index:", numbers.index(value))

print("\nPositive numbers:", positive)
print("Negative numbers:", negative)
```

```python
numbers = []

for count in range(20):
    value = int(input(f"Enter number {count + 1}: "))
    numbers.append(value)

up = sorted(numbers)
down = sorted(numbers, reverse=True)

print("Ascending:", up)
print("Descending:", down)
print("Total elements:", len(numbers))
```

```python
first = []
second = []

print("Enter 5 values for the first list:")
for _ in range(5):
    first.append(int(input()))

print("Enter 5 values for the second list:")
for _ in range(5):
    second.append(int(input()))

merged = first + second

print("First list:", first)
print("Second list:", second)
print("Combined list:", merged)
```

```python
text = input("Enter a phrase: ")
words = text.split()

print("Words:", words)

initials = ""
for word in words:
    initials += word[0].upper()

print("Initials:", initials)
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

number = int(input("Enter month number (1-12): "))

if 1 <= number <= 12:
    print("Selected month:", months[number - 1])
else:
    print("Invalid month number")
```

```python
names = []
numbers = []

for count in range(5):
    names.append(input(f"Enter student {count + 1}: "))

for count in range(5):
    numbers.append(int(input(f"Enter number for student {count + 1}: ")))

student_data = {}

for index in range(5):
    student_data[names[index]] = numbers[index]

print("Student dictionary:", student_data)
```

```python
values = []

for count in range(5):
    values.append(int(input(f"Enter value {count + 1}: ")))

print("Original list:", values)

unique_values = []

for value in values:
    if value not in unique_values:
        unique_values.append(value)
    else:
        print(f"{value} is repeated.")

print("Updated list:", unique_values)
```

```python
students = []
numbers = []

for count in range(5):
    students.append(input(f"Enter student {count + 1}: "))

for count in range(5):
    numbers.append(int(input(f"Enter number for {students[count]}: ")))

student_data = dict(zip(students, numbers))

print("Student dictionary:", student_data)
```
