```python
numbers = (10, 20, 30, 40, 50, 60, 70, 80, 90)

print("Tuple elements : ", numbers)
print("fourth element from the first : ", numbers[3])
print("fourth element from the last : ", numbers[-4])
```

```python
numbers_list = [10, 20, 30, 40, 50]
numbers_tuple = tuple(numbers_list)
print("list: ", numbers_list)
print("tuple: ", numbers_tuple)
```

```python
numbers = (10, 20, 30, 40, 50)
element = int(input("enter your element: "))
if element in numbers:
    print("index of the number is : ", numbers.index(element))
else:
    print("element not found")
```
